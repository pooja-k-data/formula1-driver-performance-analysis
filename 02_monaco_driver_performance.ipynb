```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data/monaco_2024_laps_clean.csv")
df.head()
```

```python
top_drivers = ['VER', 'LEC', 'NOR', 'SAI', 'PER']
```

```python
driver_df = df[df['Driver'].isin(top_drivers)]
```

```python
driver_df = driver_df[
    (driver_df['LapTimeSeconds'] > 70) &
    (driver_df['LapTimeSeconds'] < 100)
]
```

```python
plt.figure(figsize=(12,6))

for driver in top_drivers:
    subset = driver_df[driver_df['Driver'] == driver]
    plt.plot(
        subset['LapNumber'],
        subset['LapTimeSeconds'],
        label=driver
    )

plt.xlabel("Lap Number")
plt.ylabel("Lap Time (seconds)")
plt.title("Monaco GP 2024 — Lap Time Comparison")
plt.legend()
plt.grid(True)
plt.show()
```
