# Formula 1 Data Analytics Project

### Monaco Grand Prix 2024; Lap-Level Performance Analysis

📌 Project Overview

This project analyzes lap-level race data from the 2024 Monaco Grand Prix using Python and the FastF1 library. The goal is to explore driver performance, lap consistency, and race dynamics on one of Formula 1’s most technically demanding circuits.

The project is structured to mirror a real-world data analytics workflow:

Data ingestion from an external API

Data cleaning and transformation

Exploratory analysis and visualization

Reusable datasets for BI tools (Power BI / Tableau)

🎯 Key Questions

How consistent were top drivers throughout the Monaco GP?

Where do lap time spikes occur, and why?

How does race pace evolve over stints?

Which drivers show the strongest lap-time stability?

🧰 Tech Stack

Python

FastF1 (official F1 timing data)

Pandas (data cleaning & transformation)

Matplotlib (visualization)

Jupyter Notebook

(Phase 3) Power BI

🧪 Phase Breakdown
Phase 1 — Data Loading & Cleaning

Loaded Monaco GP 2024 race session using FastF1

Extracted lap-level timing data

Converted lap times to numeric seconds

Selected analysis-ready columns

Exported a clean dataset for reuse

📄 Output: data/monaco_2024_laps_clean.csv

Phase 2 — Driver Performance Analysis

Compare lap times of top drivers

Analyze lap-by-lap pace evolution

Identify outliers (pit stops, traffic, incidents)

Visualize performance trends

📊 Output: Driver comparison plots and insights

Phase 3 — BI Dashboard (Optional)

Import cleaned dataset into Power BI

Create interactive dashboards

Compare drivers, stints, and lap distributions
