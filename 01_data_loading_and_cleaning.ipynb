# Phase 1: Data Loading & Cleaning
## Monaco Grand Prix 2024

This notebook loads lap-level timing data from the 2024 Monaco Grand Prix
using the FastF1 library, cleans the data, and exports a reusable dataset
for further analysis and visualization.


```python
import fastf1
import pandas as pd
import os

os.makedirs("cache", exist_ok=True)
fastf1.Cache.enable_cache("cache")
```

```python
session = fastf1.get_session(2024, "Monaco", "R")
session.load()
```

```python
laps = session.laps
```

```python
laps.head()
```

```python
laps_clean = laps[
    (laps['IsAccurate'] == True) &
    (laps['LapTime'].notna())
].copy()
```

```python
laps_clean['LapTimeSeconds'] = laps_clean['LapTime'].dt.total_seconds()
```

```python
laps_clean = laps_clean[[
    'Driver',
    'LapNumber',
    'LapTimeSeconds',
    'Stint',
    'Compound'
]]
```

```python
import os
os.makedirs("data", exist_ok=True)
```

```python
import os

os.makedirs("data/raw", exist_ok=True)
```

```python
# Save raw laps exactly as loaded from FastF1
laps = session.laps.copy()

laps.to_csv(
    "data/raw/monaco_2024_laps_raw.csv",
    index=False
)
```
